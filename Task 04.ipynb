```python
import numpy as np

import matplotlib.pyplot as plt

num_col = 'radius_mean'
cat_col = 'diagnosis'

# Gaussian-like distribution for a numerical column
values = train[num_col].dropna()
mu = values.mean()
sigma = values.std(ddof=1)

x = np.linspace(values.min(), values.max(), 200)
gaussian = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(values, bins=20, density=True, alpha=0.7, color='steelblue', edgecolor='black')
axes[0].plot(x, gaussian, 'r-', linewidth=2, label='Gaussian fit')
axes[0].set_title(f'{num_col} Distribution')
axes[0].set_xlabel(num_col)
axes[0].set_ylabel('Density')
axes[0].legend()

# Bar chart distribution for a categorical column
counts = train[cat_col].value_counts()
axes[1].bar(counts.index, counts.values, color=['tab:blue', 'tab:orange'])
axes[1].set_title(f'{cat_col} Distribution')
axes[1].set_xlabel(cat_col)
axes[1].set_ylabel('Count')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report,
)

CSV_PATH = None  
RANDOM_STATE = 42


# =============================================================================
# 1. FEATURE ENGINEERING
# =============================================================================
BASE_FEATURES = [
    "radius", "texture", "perimeter", "area", "smoothness",
    "compactness", "concavity", "concave points", "symmetry", "fractal_dimension",
]


def _col(base, suffix):
    return f"{base}_{suffix}"


def safe_divide(numerator, denominator, eps=1e-6):
    return numerator / (denominator.replace(0, np.nan).abs() + eps)


def encode_target(df, target_col="diagnosis"):
    df = df.copy()
    if not pd.api.types.is_numeric_dtype(df[target_col]):
        df[target_col] = df[target_col].map({"B": 0, "M": 1}).astype(int)
    return df


def winsorize_columns(df, cols, lower_q=0.01, upper_q=0.99):
    df = df.copy()
    for c in cols:
        if c not in df.columns:
            continue
        lo, hi = df[c].quantile(lower_q), df[c].quantile(upper_q)
        df[c] = df[c].clip(lower=lo, upper=hi)
    return df


def add_triplet_ratio_features(df, base_features=BASE_FEATURES):
    df = df.copy()
    for base in base_features:
        mean_c, se_c, worst_c = _col(base, "mean"), _col(base, "se"), _col(base, "worst")
        if not all(c in df.columns for c in (mean_c, se_c, worst_c)):
            continue
        tag = base.replace(" ", "_")
        df[f"{tag}_worst_to_mean_ratio"] = safe_divide(df[worst_c], df[mean_c])
        df[f"{tag}_worst_minus_mean"] = df[worst_c] - df[mean_c]
        df[f"{tag}_se_to_mean_ratio"] = safe_divide(df[se_c], df[mean_c])
        df[f"{tag}_se_to_worst_ratio"] = safe_divide(df[se_c], df[worst_c])
    return df


def add_geometry_features(df):
    df = df.copy()
    for suffix in ["mean", "worst"]:
        radius, perimeter, area = _col("radius", suffix), _col("perimeter", suffix), _col("area", suffix)
        concavity, concave_pts = _col("concavity", suffix), _col("concave points", suffix)
        fractal_dim = _col("fractal_dimension", suffix)
        smoothness, compactness = _col("smoothness", suffix), _col("compactness", suffix)
        present = lambda *cs: all(c in df.columns for c in cs)

        if present(perimeter, radius):
            df[f"circularity_{suffix}"] = safe_divide(df[perimeter], 2 * np.pi * df[radius])
        if present(area, radius):
            df[f"area_circle_ratio_{suffix}"] = safe_divide(df[area], np.pi * df[radius] ** 2)
        if present(concave_pts, concavity, fractal_dim):
            df[f"irregularity_index_{suffix}"] = df[concave_pts] + df[concavity] + df[fractal_dim]
        if present(concave_pts, perimeter):
            df[f"concave_pts_density_{suffix}"] = safe_divide(df[concave_pts], df[perimeter])
        if present(compactness, smoothness):
            df[f"compactness_smoothness_interaction_{suffix}"] = df[compactness] * df[smoothness]
    return df


def engineer_features(df, target_col="diagnosis", winsorize_se=True, drop_id=True):
    df = df.copy()
    if drop_id and "id" in df.columns:
        df = df.drop(columns=["id"])
    df = encode_target(df, target_col=target_col)
    if winsorize_se:
        se_cols = [c for c in df.columns if c.endswith("_se")]
        df = winsorize_columns(df, se_cols)
    df = add_triplet_ratio_features(df)
    df = add_geometry_features(df)
    df = df.replace([np.inf, -np.inf], np.nan)
    return df


# =============================================================================
# 2. DATA LOADING
# =============================================================================
def load_data(csv_path=None):
    if csv_path:
        return pd.read_csv(csv_path)

    data = load_breast_cancer()
    df = pd.DataFrame(data.data, columns=data.feature_names)
    rename_map = {}
    for c in df.columns:
        parts = c.split()
        if parts[0] == "mean":
            rename_map[c] = f"{' '.join(parts[1:])}_mean"
        elif parts[0] == "worst":
            rename_map[c] = f"{' '.join(parts[1:])}_worst"
        elif parts[-1] == "error":
            rename_map[c] = f"{' '.join(parts[:-1])}_se"
    df = df.rename(columns=rename_map)
    df["diagnosis"] = np.where(data.target == 0, "M", "B")
    return df


raw_df = load_data(CSV_PATH)
fe_df = engineer_features(raw_df)

X = fe_df.drop(columns=["diagnosis"])
y = fe_df["diagnosis"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=RANDOM_STATE
)

print(f"Train shape: {X_train.shape}, Test shape: {X_test.shape}\n")


# =============================================================================
# 3. LOGISTIC REGRESSION
# =============================================================================
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train.fillna(X_train.median()))
X_test_scaled = scaler.transform(X_test.fillna(X_train.median()))

model = LogisticRegression(max_iter=5000, random_state=RANDOM_STATE)
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)
y_proba = model.predict_proba(X_test_scaled)[:, 1]


# =============================================================================
# 4. CLASSIFICATION METRICS
# =============================================================================
print("Accuracy :", round(accuracy_score(y_test, y_pred), 4))
print("Precision:", round(precision_score(y_test, y_pred), 4))
print("Recall   :", round(recall_score(y_test, y_pred), 4))
print("F1 Score :", round(f1_score(y_test, y_pred), 4))
print("ROC-AUC  :", round(roc_auc_score(y_test, y_proba), 4))

print("\nConfusion Matrix ([[TN, FP], [FN, TP]]):")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=["Benign (0)", "Malignant (1)"]))
```
